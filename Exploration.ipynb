```python
manque = data_ch.isnull().mean()*100
manque = manque.sort_values(ascending= False)
manque
```

```python
# Dictionnaire de renommage avec des noms explicites
traductions = {
    'HH1': 'id_grappe',
    'HH2': 'id_menage',
    'AN3': 'id_ligne_mere',
    'AN5': 'age_mere',
    'HL4': 'sexe',
    'CAGE': 'age_enfant_mois',
    'melevel': 'niveau_instruction_mere',
    'BD3': 'jamais_allaite',
    'CA1': 'fievre_recente',
    'CA14': 'toux_ou_ira',
    'HAZ2': 'zscore_taille_age',
    'WAZ2': 'zscore_poids_age',
    'WHZ2': 'zscore_poids_taille',
    'HAZFLAG': 'flag_taille_age',
    'WAZFLAG': 'flag_poids_age',
    'WHZFLAG': 'flag_poids_taille',
    'FLAG': 'flag_anthropometrie_globale'
}
# Renommage des colonnes 
data_ch = data_ch.rename(columns=traductions)
```

```python
data_ch.columns # Relecture de chaque colonnes
```

```python
data_ch.head()
```

```python
df_ch = data_ch.iloc[:, 4:]  # Abandon des id
df_ch.head()
```

```python
plt.figure(figsize= (18, 7))
sns.heatmap(df_ch.corr(numeric_only= True), annot = True, cmap = "Set2", alpha= 0.8)
plt.title("Table de correlation entre les variables predictives\n", c = "red", size = 18)
plt.xticks(rotation = 70)
plt.show()
```

- `Les correlations entre plusieur variables sont assez elevee d'autre trop faibles `
